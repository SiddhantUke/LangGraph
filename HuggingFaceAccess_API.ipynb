This code is checking which AI models your Hugging Face account can currently access through the Hugging Face Router API.

In your situation, this was useful because you were getting:

`model_not_supported`

So instead of guessing model names, we asked Hugging Face:

`"Which models are available to me?"`

```python
import requests
import os

headers = {
    "Authorization": f"Bearer {os.environ['HF_TOKEN']}"
}

response = requests.get(
    "https://router.huggingface.co/v1/models",
    headers=headers
)

data = response.json()

for model in data["data"][:20]:
    print(model["id"])
```

```python
import os

from langchain_huggingface import (
    HuggingFaceEndpoint,
    ChatHuggingFace,
    HuggingFaceEmbeddings
)

# -------------------------
# 1. LLM
# -------------------------

endpoint = HuggingFaceEndpoint(
    repo_id="openai/gpt-oss-20b",
    huggingfacehub_api_token=os.environ["HF_TOKEN"],
    max_new_tokens=500,
    temperature=0.1
)

chat_model = ChatHuggingFace(llm=endpoint)


# -------------------------
# 2. Generate response
# -------------------------

response = chat_model.invoke(
    "What is Machine Learning?"
)

text = response.content

print("LLM RESPONSE:")
print(text)


# -------------------------
# 3. Embedding model
# -------------------------

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)


# -------------------------
# 4. Convert response → vector
# -------------------------

vector = embeddings.embed_query(text)

print("\nEMBEDDING:")
print(vector)

print("\nVECTOR DIMENSION:")
print(len(vector))
```
